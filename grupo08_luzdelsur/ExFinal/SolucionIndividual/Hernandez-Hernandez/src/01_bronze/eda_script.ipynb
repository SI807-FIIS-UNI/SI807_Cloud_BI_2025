```python
# --- 1. CONFIGURACIÓN INICIAL ---
storage_account = "datalakemichi1"
container = "bronce"
base_path = f"abfss://{container}@{storage_account}.dfs.core.windows.net/raw"

# Lista de tus archivos subidos
files = {
    "city_day": "city_day.csv",
    "city_hour": "city_hour.csv",
    "station_day": "station_day.csv",
    "station_hour": "station_hour.csv",
    "stations": "stations.csv"
}

# --- 2. PROCESO DE INGESTA Y EDA ---
for table_name, file_name in files.items():
    print(f"--- Procesando: {file_name} ---")
    
    # A. Leer el CSV desde el Data Lake
    file_path = f"{base_path}/{file_name}"
    
    # Usamos inferSchema=True para que detecte números y fechas automáticamente
    df = spark.read.format("csv") \
        .option("header", "true") \
        .option("inferSchema", "true") \
        .load(file_path)
    
    # B. EDA Mínimo (Requisito del examen)
    print(f"Registros encontrados: {df.count()}")
    df.printSchema() # Muestra la estructura de columnas
    
    # C. Guardar como Tabla Delta (Capa Bronce)
    # Guardamos como 'bronze_nombretabla' para mantener orden
    bronze_table_name = f"bronze_{table_name}"
    
    df.write.format("delta") \
        .mode("overwrite") \
        .option("mergeSchema", "true") \
        .saveAsTable(bronze_table_name)
    
    print(f"✅ Tabla creada exitosamente: {bronze_table_name}\n")

print("¡INGESTA Y EDA COMPLETADOS!")
```
