# CARGA A BIGQUERY

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("carga_bigquery").getOrCreate()

# 🧾 Proyecto y dataset de destino en BigQuery
project_id = "shaped-icon-478404-p0"
dataset_id = "retail_transaction"


dim_tienda = spark.read.parquet("gs://retail-transactions-final/refined/dimensiones/dim_tienda")

dim_tienda.write \
    .format("bigquery") \
    .option("table", f"{project_id}:{dataset_id}.dim_tienda") \
    .mode("overwrite") \
    .save()

dim_cliente = spark.read.parquet("gs://retail-transactions-final/refined/dimensiones/dim_cliente")

dim_cliente.write \
    .format("bigquery") \
    .option("table", f"{project_id}:{dataset_id}.dim_cliente") \
    .mode("overwrite") \
    .save()


dim_tiempo = spark.read.parquet("gs://retail-transactions-final/refined/dimensiones/dim_tiempo")

dim_tiempo.write \
    .format("bigquery") \
    .option("table", f"{project_id}:{dataset_id}.dim_tiempo") \
    .mode("overwrite") \
    .save()


fact_transacciones = spark.read.parquet("gs://retail-transactions-final/refined/hechos/fact_transacciones")

fact_transacciones.write \
    .format("bigquery") \
    .option("table", f"{project_id}:{dataset_id}.fact_transacciones") \
    .mode("overwrite") \
    .save()

print("✅ Todas las tablas cargadas correctamente a BigQuery.")
```
