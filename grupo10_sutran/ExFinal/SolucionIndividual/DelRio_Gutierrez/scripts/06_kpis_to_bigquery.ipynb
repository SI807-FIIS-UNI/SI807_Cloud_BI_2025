```python
# Spark config
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("kpis_to_bigquery").getOrCreate()

project_id = "shaped-icon-478404-p0"
dataset_id = "retail_transaction"
temp_bucket = "retail-transactions-final"

# Cargar y subir cada KPI

# 1. Ticket Promedio
ticket = spark.read.parquet("gs://retail-transactions-final/kpis/ticket_promedio")
ticket.write \
    .format("bigquery") \
    .option("table", f"{project_id}:{dataset_id}.kpi_ticket_promedio") \
    .option("temporaryGcsBucket", temp_bucket) \
    .mode("overwrite") \
    .save()

# 2. Horas Pico
horas = spark.read.parquet("gs://retail-transactions-final/kpis/horas_pico")
horas.write \
    .format("bigquery") \
    .option("table", f"{project_id}:{dataset_id}.kpi_horas_pico") \
    .option("temporaryGcsBucket", temp_bucket) \
    .mode("overwrite") \
    .save()

# 3. Frecuencia de compra
frecuencia = spark.read.parquet("gs://retail-transactions-final/kpis/frecuencia_compras")
frecuencia.write \
    .format("bigquery") \
    .option("table", f"{project_id}:{dataset_id}.kpi_frecuencia_compras") \
    .option("temporaryGcsBucket", temp_bucket) \
    .mode("overwrite") \
    .save()

# 4. Promociones
promos = spark.read.parquet("gs://retail-transactions-final/kpis/promociones")
promos.write \
    .format("bigquery") \
    .option("table", f"{project_id}:{dataset_id}.kpi_promociones") \
    .option("temporaryGcsBucket", temp_bucket) \
    .mode("overwrite") \
    .save()

print("✅ KPIs cargados a BigQuery. ¡Listos para Looker Studio!")
```
